```python
df = pd.read_csv('books_data.csv')

print(df.head())
print(df.shape)
```

```python
df['Price'] = pd.to_numeric(
    df['Price'].astype(str).str.replace(r'[^0-9.]', '', regex=True),
    errors='coerce'
)

rating_map = {
    'One': 1,
    'Two': 2,
    'Three': 3,
    'Four': 4,
    'Five': 5
}

df['Rating_Number'] = df['Rating'].map(rating_map)

print(df.head())
print(df.dtypes)
```

```python
avg_price = df.groupby('Rating_Number')['Price'].mean()

plt.figure(figsize=(8,5))
plt.bar(avg_price.index, avg_price.values)

plt.title('Average Book Price by Rating')
plt.xlabel('Book Rating')
plt.ylabel('Average Price (£)')
plt.xticks([1, 2, 3, 4, 5])

plt.show()
```

```python
rating_count = df['Rating'].value_counts().reindex(
    ['One', 'Two', 'Three', 'Four', 'Five']
)

plt.figure(figsize=(8,5))
plt.bar(rating_count.index, rating_count.values)

plt.title('Number of Books by Rating')
plt.xlabel('Book Rating')
plt.ylabel('Number of Books')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Price'], bins=8, edgecolor='black')

plt.title('Distribution of Book Prices')
plt.xlabel('Book Price (£)')
plt.ylabel('Number of Books')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Rating_Number'], df['Price'])

plt.title('Book Price vs Rating')
plt.xlabel('Book Rating')
plt.ylabel('Book Price (£)')

plt.xticks([1, 2, 3, 4, 5])
plt.grid(True)

plt.show()
```

```python
# 1. Average Book Price by Rating
avg_price = df.groupby('Rating_Number')['Price'].mean()

plt.figure(figsize=(8,5))
plt.bar(avg_price.index, avg_price.values)
plt.title('Average Book Price by Rating')
plt.xlabel('Book Rating')
plt.ylabel('Average Price (£)')
plt.savefig('average_price_by_rating.png', bbox_inches='tight')
plt.show()

# 2. Number of Books by Rating
rating_count = df['Rating'].value_counts().reindex(
    ['One', 'Two', 'Three', 'Four', 'Five']
)

plt.figure(figsize=(8,5))
plt.bar(rating_count.index, rating_count.values)
plt.title('Number of Books by Rating')
plt.xlabel('Book Rating')
plt.ylabel('Number of Books')
plt.savefig('books_by_rating.png', bbox_inches='tight')
plt.show()

# 3. Book Price Distribution
plt.figure(figsize=(8,5))
plt.hist(df['Price'], bins=8, edgecolor='black')
plt.title('Distribution of Book Prices')
plt.xlabel('Book Price (£)')
plt.ylabel('Number of Books')
plt.savefig('book_price_distribution.png', bbox_inches='tight')
plt.show()

# 4. Book Price vs Rating
plt.figure(figsize=(8,5))
plt.scatter(df['Rating_Number'], df['Price'])
plt.title('Book Price vs Rating')
plt.xlabel('Book Rating')
plt.ylabel('Book Price (£)')
plt.xticks([1,2,3,4,5])
plt.grid(True)
plt.savefig('book_price_vs_rating.png', bbox_inches='tight')
plt.show()
```
